## 一个 30 秒的热身：注意力分数矩阵

后面每个模块都会反复用到 `softmax` 与缩放点积注意力。先把它们写出来当作全课的基础件。

```python
def softmax(x, axis=-1):
    x = x - x.max(axis=axis, keepdims=True)
    e = np.exp(x)
    return e / e.sum(axis=axis, keepdims=True)

def attention(Q, K, V, causal=True):
    d = Q.shape[-1]
    scores = Q @ K.T / np.sqrt(d)          # (T, T)
    if causal:
        T = scores.shape[0]
        mask = np.triu(np.ones((T, T), dtype=bool), k=1)
        scores = np.where(mask, -np.inf, scores)
    A = softmax(scores, axis=-1)
    return A @ V, A

rng = np.random.default_rng(0)
T, d = 5, 8
Q = K = V = rng.standard_normal((T, d))
out, A = attention(Q, K, V)
print('注意力矩阵行和应为 1：', np.allclose(A.sum(1), 1))
print('因果掩码下右上角应为 0：', np.allclose(np.triu(A, 1).sum(), A[np.triu_indices(T,1)].sum()))
print('输出形状', out.shape)
```
